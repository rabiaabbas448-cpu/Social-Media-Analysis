Social Media Addiction & Student Behavior Analysis

This project analyzes data from 705 students to understand how social media usage affects sleep, mental health, and academic performance.

**Dataset:** Students Social Media Addiction (Kaggle)
**Tools:** Python, Pandas, Matplotlib, Seaborn




## 1. Loading the Dataset


```python
from google.colab import files
files.upload()
```

```python
import pandas as pd

df = pd.read_csv("dataset.csv.csv")
df.head()
```

## 2. Understanding the Data
The dataset has 705 rows and 13 columns, with no missing values.

```python
print(df.shape)
print(df.isnull().sum())
df.info()
```

## 3. Social Media Usage vs Sleep

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Avg_Daily_Usage_Hours", y="Sleep_Hours_Per_Night")
plt.title("Social Media Usage vs Sleep Hours")
plt.xlabel("Daily Usage (Hours)")
plt.ylabel("Sleep (Hours)")
plt.show()
```

Correlation between usage and sleep: **-0.79** (strong negative relationship).

```python
df["Avg_Daily_Usage_Hours"].corr(df["Sleep_Hours_Per_Night"])
```

## 4. Correlation Heatmap

```python
plt.figure(figsize=(9, 6))
sns.heatmap(df.drop(columns="Student_ID").corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

## 5. Addiction Score by Platform

```python
platform_avg = df.groupby("Most_Used_Platform")["Addicted_Score"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=platform_avg.index, y=platform_avg.values)
plt.title("Average Addiction Score by Platform")
plt.xlabel("Platform")
plt.ylabel("Average Addicted Score")
plt.xticks(rotation=45)
plt.show()
```

## 6. Impact on Academic Performance

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Affects_Academic_Performance")
plt.title("Does Social Media Affect Academic Performance?")
plt.xlabel("Affects Academic Performance")
plt.ylabel("Number of Students")
plt.show()
```

## 7. Conclusion

Key findings from the analysis of 705 students:

- **Sleep:** Higher daily social media usage is strongly linked to fewer hours of sleep (correlation: -0.79).
- **Mental health:** Addiction score has a very strong negative relationship with mental health score (-0.95) and a strong positive relationship with conflicts over social media (+0.93).
- **Platforms:** WhatsApp, Snapchat, and TikTok users have the highest average addiction scores (about 7.4-7.5), while LinkedIn and LINE users have the lowest.
- **Academic impact:** About 64% of students (453 out of 705) say social media affects their academic performance. These students also use social media more (5.5 hours/day vs 3.8 hours/day).

**Note:** These results show relationships, not proof of cause and effect. Also, some platforms (like Snapchat) have very few students in the data, so their averages are less reliable.
